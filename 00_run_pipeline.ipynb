```python
def run(cmd, name="cmd"):
    ts = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    log_path = LOG_DIR / f"{ts}_{name}.txt"

    print("\n" + "="*80)
    print(">>", cmd)

    r = subprocess.run(cmd, shell=True, text=True, capture_output=True)
    out = (r.stdout or "") + ("\n" + (r.stderr or ""))
    log_path.write_text(out, encoding="utf-8", errors="ignore")

    print(f"[LOG] {log_path}")
    print(out[:2000])

    if r.returncode != 0:
        raise RuntimeError(f"Falhou (code={r.returncode}). Veja o log: {log_path}")

    return out
```

```python
from pathlib import Path
import subprocess
import datetime

def run_live(cmd, name="cmd"):
    ts = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    log_path = LOG_DIR / f"{ts}_{name}.txt"

    print("\n" + "="*80)
    print(">>", cmd)
    print(f"[LOG] {log_path}")

    with open(log_path, "w", encoding="utf-8", errors="ignore") as f:
        p = subprocess.Popen(
            cmd,
            shell=True,
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            text=True,
            bufsize=1,
            universal_newlines=True,
        )
        for line in p.stdout:
            print(line, end="")   # aparece ao vivo no notebook
            f.write(line)         # salva no log ao mesmo tempo

        code = p.wait()

    if code != 0:
        raise RuntimeError(f"Falhou (code={code}). Veja o log: {log_path}")

    return str(log_path)
```
